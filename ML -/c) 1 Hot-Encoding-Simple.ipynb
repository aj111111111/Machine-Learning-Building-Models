```python
import pandas as pd
from sklearn.linear_model import LinearRegression

# ==========================================
# 1. डेटा लोड करना
# ==========================================
df = pd.read_csv('homeprices.csv')
print('-> 1. Original DataFrame (डेटा):')
print(df)
print('=' * 60)

# ==========================================
# 2. One-Hot Encoding (Pandas get_dummies का उपयोग करके)
# ==========================================
# यह 'town' कॉलम को बाइनरी (0 और 1) कॉलम में बदल देगा
dummies = pd.get_dummies(df['town'], dtype=int)
merged = pd.concat([df, dummies], axis=1)
encoded_df = merged.drop(['town'], axis=1)

print('-> 2. Encoded DataFrame (One-Hot Encoding के बाद):')
print(encoded_df)
print('=' * 60)

```

```python
# ==========================================
# 3. Features (X) और Target (y) को अलग करना
# ==========================================
# यहाँ हम price को प्रेडिक्ट करना चाहते हैं, इसलिए y = price और बाकी सब X = features हैं।
X = encoded_df.drop(['price'], axis=1)
y = encoded_df['price']

print('-> 3. Training Features (X_train - जिन डेटा से मॉडल सीखेगा):')
print(X)
print('\n-> Target Variable (y_train - जिसे मॉडल को प्रेडिक्ट करना है):')
print(y)
print('=' * 60)

# ==========================================
# 4. Model Training (मॉडल को ट्रेन करना)
# ==========================================
model = LinearRegression()
model.fit(X, y)  # यहाँ मॉडल पूरी तरह ट्रेन हो गया

print('-> 4. Model Training Successful! (मॉडल सीख चुका है)')
print(f'   - Model Coefficients (वजन): {model.coef_}')
print(f'   - Model Intercept (इंटरसेप्ट): {model.intercept_}')
print('=' * 60)
```

```python
# ==========================================
# 5. Prediction (नए घर की कीमत का अनुमान लगाना)
# ==========================================
# मान लीजिए हमें 3400 sq-ft के घर की कीमत निकालनी है जो 'robinsville' शहर में है।
# कॉलम का क्रम वही होना चाहिए जो X में है: [area, monroe township, robinsville, west windsor]
# robinsville के लिए robinsville वाले कॉलम में 1 और बाकी में 0 होगा।

# नया डेटा पॉइंट बनाना
new_home_data = [[3400, 0, 1, 0]]  # 3400 sq-ft, Robinsville

predicted_price = model.predict(new_home_data)

print('-> 5. Prediction Result:')
print(
    f'   - 3400 sq-ft घर (शहर: Robinsville) की अनुमानित कीमत (Predicted Price):'
    f' ₹{predicted_price[0]:,.2f}'
)
```
